## 8. License status

```python
bcn['license'].isna().sum()
```

```python
bcn['license'].isna().mean()
```

```python
bcn["license"].value_counts().head(20)
```

The `license` field is free text written by hosts. In Barcelona it mixes two registries (regional and national), contains HTML line breaks (`<br />`) and different code formats (`HUTB-` for tourist apartments, `HB-` for hotels, `AJ` for hostels). 36% of listings have no license information at all.

To compare the three cities with the same criteria, listings are grouped into three categories:

- **number_declared**: the host provides a registration number (any text containing a digit)
- **exempt**: the host declares the listing as exempt (e.g. seasonal rental), with no registration number
- **missing**: no license information

⚠️ Registration numbers are self-declared and **not validated** against official registries. Some look like placeholders (for example `HUTB-123456` appears in 12 listings), so `number_declared` should be read as an upper bound of properly licensed listings.

```python
bcn['license_status'] = 'missing'

bcn.loc[bcn['license'].str.contains('Exempt', na=False), 'license_status'] = 'exempt'

bcn.loc[bcn['license'].str.contains(r'\d', na=False), 'license_status'] = 'number_declared'

bcn["license_status"].value_counts(normalize=True)
```

**Key finding:** 35.9% of listings in Barcelona show no license information, and only 59.6% declare a registration number, which is an upper bound since numbers are not validated. 4.5% declare themselves exempt, mostly as seasonal rentals.

# Part 2 - Cleaning pipeline

## 9. Cleaning function

All cleaning steps decided in Part 1 are grouped into a single function, so the same logic can be applied to the three cities consistently.

```python
def clean_listings(df):
    # 1. Keep only the selected columns
    df = df[cols].copy()
    
    # 2. Price: remove "$" and "," and convert to numeric
    df['price'] = df['price'].str.replace('$','')
    df['price'] = df['price'].str.replace(',','')
    df['price'] = pd.to_numeric(df['price'])

    # 3. Flag price outliers (above the 99th percentile of the city)
    p99 = df['price'].quantile(0.99)
    df['is_price_outlier'] = df['price'] > p99

    # 4. Superhost: "t"/"f" -> True/False
    df['host_is_superhost'] = df['host_is_superhost'].map({'t': True, 'f': False})

    # 5. Last review: text -> date
    df['last_review'] = pd.to_datetime(df['last_review'])

    # 6. Host type: single / multi / professional
    df['host_type'] = pd.cut(
        df['calculated_host_listings_count'],
        bins=[0, 1, 9, float("inf")],
        labels=['single', 'multi', 'professional'])

    # 7. License status: missing / exempt / number_declared / business_info
    df['license_status'] = 'missing'
    df.loc[df['license'].str.contains('Exempt', na=False), 'license_status'] = 'exempt'
    df.loc[df['license'].str.contains(r'\d', na=False), 'license_status'] = 'number_declared'
    df.loc[df['license'].str.contains('Legal entity name|First name and Last name', case=False, na=False), 'license_status'] = 'identity_info'

    return df
    
```

```python
bcn_raw = pd.read_csv(DATA_PATH + 'barcelona-listings.csv.gz')
bcn_clean = clean_listings(bcn_raw)
bcn_clean['license_status'].value_counts(normalize=True)
```

## 10. Apply to all cities

### 10.1 Check license format in Berlin and Amsterdam

The cleaning rules were designed by exploring Barcelona only. Before applying the function to the other cities, we check that the `license` field follows a similar format, since each city has its own registration system and the rules (digit → `number_declared`, "Exempt" → `exempt`) may need adjusting.

```python
ber_raw = pd.read_csv(DATA_PATH + 'berlin-listings.csv.gz')

# Count license formats without displaying personal data
lic = ber_raw['license']
pd.Series({
    'missing': lic.isna().sum(),
    'exempt': lic.str.contains('Exempt', na=False).sum(),
    'registration number (XX/Z/XX/...)': lic.str.contains('/Z/', na=False).sum(),
    'legal entity details': lic.str.contains('Legal entity name', na=False).sum(),
    'individual details': lic.str.contains('First name and Last name', na=False).sum(),
})
```

Berlin uses a different format from Barcelona. Besides official registration numbers, many hosts publish identity details instead: company details ("Legal entity name...") or personal details ("First name and Last name..."). These texts contain digits (postcodes, street numbers), so the original digit rule would wrongly classify them as `number_declared`.

To avoid displaying personal data, raw license texts from Berlin are not shown in this notebook.

```python
ams_raw = pd.read_csv(DATA_PATH + 'amsterdam-listings.csv.gz')
ams_raw['license'].value_counts(dropna=False).head(20)
```

Amsterdam fits the original rules: "Exempt", empty values, or official registration numbers (almost all starting with `0363`, the municipality code). Some numbers look like placeholders (`ABCD 1234 AB12 89EF A0F9` in 10 listings).

### 10.2 Updated license rules

A fourth category, `identity_info`, is added to the cleaning function (section 9) for listings showing identity details instead of a registration number. Final rules, applied in this order (each rule overrides the previous one):

1. `missing`: no license information
2. `exempt`: contains "Exempt"
3. `number_declared`: contains a digit
4. `identity_info`: contains "Legal entity name" or "First name and Last name"

```python
ber_clean = clean_listings(ber_raw)
ber_clean['license_status'].value_counts(dropna=False)
```

**Result:** After adding the `identity_info` category, 34.9% of Berlin listings declare a registration number, 31.9% show identity details instead (company or personal), 32.6% have no license information and 0.6% are exempt. Without this category, the share of listings with a registration number would have been overstated (47.8% instead of 34.9%).

### 10.3 Clean and combine the three cities

The cleaning function is applied to each city, a `city` column is added, and the three datasets are combined into a single table.

```python
bcn_clean = clean_listings(bcn_raw)
ber_clean = clean_listings(ber_raw)
ams_clean = clean_listings(ams_raw)

bcn_clean['city'] = 'Barcelona'
ber_clean['city'] = 'Berlin'
ams_clean['city'] = 'Amsterdam'

listings = pd.concat([bcn_clean, ber_clean, ams_clean], ignore_index=True)

listings['city'].value_counts()
```

In Barcelona, more than a third of listings (35.9%) show no license information, and in Berlin only 34.9% declare a registration number, while 31.9% show identity details instead.

Note: declared numbers are not validated (some are placeholders), so these shares are upper bounds of properly licensed listings.

However, the share of active listings without price differs between cities: 27.3% in Amsterdam, 13.2% in Berlin and 8.7% in Barcelona. Amsterdam's price and revenue figures should therefore be interpreted with more caution.
